```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd
```

```python
data = pd.read_csv('owid-co2-data.csv', on_bad_lines='skip', delimiter= ',')
```

```python
data.head()
```

```python
#filter to only have year (2023)
data = data[data['year'] == 2023]

```

```python
data
```

```python
#change column name from iso_code to ISO3_CODE
data.rename(columns={'iso_code': 'ISO3_CODE'}, inplace=True)
```

```python
data
```

```python
#drop continents (any NaN inthe ISO3_CODE column)
data = data.dropna(subset=['ISO3_CODE'])
```

```python
countries = gpd.read_file('/Users/henok/Dropbox/18Henok/homework/week11_hw/CNTR_RG_60M_2024_4326.shp/CNTR_RG_60M_2024_4326.shp')
```

```python
countries
```

```python
#Combine the data with the shapefile
countries = countries.merge(data, on='ISO3_CODE', how='left')
# if column co2 is NaN or 0, replace it with 0
```

```python
countries.describe()
```

```python
#plot countries
fig, ax = plt.subplots(figsize=(10, 15))
countries.plot(column='total_ghg', cmap='coolwarm', ax=ax, legend=True) #ax=ax combines the two plots

ax.axis('off')

#add title
plt.title('Total GHG Emissions by Country in 2023', fontsize=20)

plt.show()
#plot data
```
