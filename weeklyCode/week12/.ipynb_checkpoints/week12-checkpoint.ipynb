```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('floridaHousePrices.csv', index_col='Index')
data.head()
```

```python
dtype = data.dtypes
print(dtype)
```

```python
#plot the data
fig, ax = plt.subplots(figsize=(8, 6))
plt.scatter(data.Size, data.Price/1000)
```

```python
from scipy.stats import linregress
```

```python
#fit a linear regression model
model = linregress(data.Size, data.Price)
m, b = model.slope, model.intercept

x = np.arange(500,4000,100)
yhat = m*x + b
```

```python
#sum of squared risiduals

#np.sum(data.Price - (m*data.Size + b))**2

#summing the squared residuals
ssrs = (data.Price - (m*data.Size + b))**2 
ssrs.sum()
```

```python
training = data.sample(frac=0.6)
test = data[~data.index.isin(training.index)]
```

```python
N = 5000


testErrors = np.zeros(N)
#running the model 5000 times
for i in range(N):
    training = data.sample(frac=0.6)
    test = data[~data.index.isin(training.index)]
    
    model = linregress(training.Size, training.Price)
    m, b = model.slope, model.intercept

    x = np.arange(500,4000,100)
    yhat = m*x + b
    ssrs = (data.Price - (m*data.Size + b))**2
    testErrors[i] = ssrs.sum()
```

```python
model = linregress(test.Size, test.Price)
m, b = model.slope, model.intercept

x = np.arange(500,4000,100)
yhat = m*x + b
```

```python
#plot the data
fig, ax = plt.subplots(figsize=(8, 6))
plt.scatter(data['Size'], data['Price']/1000)
plt.scatter(training['Size'], training['Price']/1000, color='red')
ax.plot(x, yhat/1000, color='black')


ssrs = (data.Price - (m*data.Size + b))**2

ssr = ssrs.sum()
ax.text(500, 300, f'SSR: {ssr: 2e}')
```
