```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('floridaHousePrices.csv', index_col='Index')
data.head()
#don't use quadrant
#don't use taxes
```

```python
data['Size2'] = data['Size'] ** 2
```

```python
#sklearn requires you to split the data into features and target
#x = data[['Size', 'Lotsize', 'Size2']] # 25896.334256595095
x = data[['Bedrooms', 'Lotsize', 'Size2', 'Size']] # 21790.008623985552
#x = data[['Size', 'Bedrooms', 'Size2']] # 24817.318033698786
y = data['Price']
```

```python
#spliting the data into training and testing sets, sklearn has a function for it
from sklearn.model_selection import train_test_split

#prepping the data in advance
#usually more data for training is better 
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3)
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape #70%, 30%
```

```python
#linregress function equivalent in sklearn
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
model.fit(x_train, y_train)
```

```python
model.coef_ #slope (m)
#for each increase in square footage, the price increases by $74
#for each increase in number of baths, the price increases by $8,418
```

```python
model.intercept_ #y-intercept (b)
```

```python
model.score(x_test, y_test) #R^2 score
```

```python
model.predict(x_test) #predicting the y values for the test set (cost prediction)
```

```python
from sklearn.metrics import mean_squared_error as mse
```

```python
np.sqrt(mse(y_test, model.predict(x_test))) #mean squared error
```
