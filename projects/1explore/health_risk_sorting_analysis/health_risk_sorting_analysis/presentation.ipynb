Bubble Sort

Time Complexity: O(n²)
Space Complexity: O(1)

```python
#import
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
```

```python
#numbers from 0 to 10
numb = list(np.random.randint(1, 10, 6))
print(numb)
```

```python
def bubble_sort(arr):
    arr = arr.copy() #doesnt change the original array
    steps = [arr.copy()] #store the steps 
    for i in range(len(arr)):
        for j in range(len(arr) - i - 1):
            if arr[j] > arr[j+1]: #swaps
                temp = arr[j]
                arr[j] = arr[j+1]
                arr[j+1] = temp 
                steps.append(arr.copy()) #stores change
    return steps
steps = bubble_sort(numb)
```

```python
def plot_sort_step(step):
    plt.figure(figsize=(8,4))
    plt.bar(range(len(steps[step])), steps[step], color='red')
    plt.title(f"Bubble Sort Step {step+1} of {len(steps)}")
    plt.xlabel("Index")
    plt.ylabel("Score")
    plt.ylim(0, max(numb)+10)
    plt.grid(True)
    plt.show()

widgets.interact(plot_sort_step, step=widgets.IntSlider(min=0, max=len(steps)-1, step=1, value=0))
```

Insertion Sort

Time Complexity: O(n²)
Space Complexity: O(1)

```python
numb #check values
```

```python
def insertion_sort_steps(arr):
    arr = arr.copy()
    steps = [arr.copy()]
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        # Shift elements that are greater than key one position ahead
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            steps.append(arr.copy())
        arr[j + 1] = key
        steps.append(arr.copy())
    return steps

steps = insertion_sort_steps(numb)
```

```python
def plot_sort_step(step):
    plt.figure(figsize=(8,4))
    plt.bar(range(len(steps[step])), steps[step], color='green')
    plt.title(f"Insertion Sort Step {step+1} of {int(len(steps))}")
    plt.xlabel("Index")
    plt.ylabel("Score")
    plt.ylim(0, max(numb)+10)
    plt.grid(True)
    plt.show()

widgets.interact(plot_sort_step, step=widgets.IntSlider(min=0, max=len(steps)-1, step=1, value=0))
```

```python
# store calcuated values
risk_scores = []

# Open file
with open('/Users/henok/Dropbox/18Henok/projects/1explore/health_risk_sorting_analysis/data/processed/processed_data.csv', 'r', newline='', encoding='utf8') as infile:
    reader = csv.DictReader(infile)
    
    for row in reader:
        # Convert and extract values. Adjust these conversions if needed.
        age = float(row['Age'])
        sleep_duration = float(row['Sleep Duration'])
        sleep_quality = float(row['Quality of Sleep'])
        physical_activity = float(row['Physical Activity Level'])
        stress_level = float(row['Stress Level'])
        heart_rate = float(row['Heart Rate'])
        daily_steps = float(row['Daily Steps'])

        # 0 Male, 1 Female.
        gender = 0   
        if row['Gender'].strip().lower() == 'female':
            gender = 1

        # Convert BMI category to a numeric value.
        bmi_category = 0 #normal
        if row['BMI Category'].strip().lower() == 'overweight':
            bmi_category = 1
        elif row['BMI Category'].strip().lower() == 'Obese':
            bmi_category = 2

        # Calculate risk score
        risk = (
            0.3 * age + 15 * abs(8 - sleep_duration) + 20 * (10 - sleep_quality) +
            0.7 * max(0, 30 - physical_activity) + 10 * stress_level + 2 * bmi_category +
            0.2 * heart_rate * (1 + 0.1 * gender) + 0.001 * max(0, 10000 - daily_steps)
        )
        
        risk_scores.append(risk)

# Print or further process the risk_scores array.
print(risk_scores)  # Print the first 10 risk scores
```

Bubble Sort

```python
def bubble_sort(arr):
    arr = arr.copy() #doesnt change the original array
    steps = [arr.copy()] #store the steps 
    n = len(arr)
    for i in range(n):
        for j in range(n - i - 1):
            if arr[j] > arr[j+1]: #swaps
                temp = arr[j]
                arr[j] = arr[j+1]
                arr[j+1] = temp
                steps.append(arr.copy()) #stores change
    return steps
steps = bubble_sort(risk_scores)
```

```python
def plot_sort_step(step):
    plt.figure(figsize=(8,4))
    plt.bar(range(len(steps[step])), steps[step], color='orange')
    plt.title(f"Bubble Sort Step {step+1} of {len(steps)}")
    plt.xlabel("Patient Index")
    plt.ylabel("Risk Score")
    plt.ylim(0, max(risk_scores)+10)
    plt.grid(True)
    plt.show()

widgets.interact(plot_sort_step, step=widgets.IntSlider(min=0, max=len(steps)-1, step=1, value=0))
```
